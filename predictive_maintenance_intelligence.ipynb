```python
# Define the input features
feature_columns = [
    "Type",
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]"
]

# X contains the input features
X = df[feature_columns]

# y contains the target
y = df["Machine failure"]

print("Feature shape:", X.shape)
print("Target shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))

print("\nTraining failure proportion:")
print(y_train.value_counts(normalize=True).mul(100).round(2))

print("\nTesting failure proportion:")
print(y_test.value_counts(normalize=True).mul(100).round(2))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline

# Identify numerical and categorical features
numeric_features = [
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]"
]

categorical_features = ["Type"]

# Preprocessing for numerical features
numeric_transformer = Pipeline(
    steps=[
        ("scaler", StandardScaler())
    ]
)

# Preprocessing for categorical features
categorical_transformer = Pipeline(
    steps=[
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]
)

# Combine both preprocessing pipelines
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("Preprocessing pipeline created successfully!")
```

```python
# Fit preprocessing on training data only
X_train_processed = preprocessor.fit_transform(X_train)

# Transform test data using the fitted preprocessor
X_test_processed = preprocessor.transform(X_test)

print("Original training shape:", X_train.shape)
print("Processed training shape:", X_train_processed.shape)

print("Original testing shape:", X_test.shape)
print("Processed testing shape:", X_test_processed.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# Create the complete ML pipeline
logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(
            max_iter=1000,
            random_state=42
        ))
    ]
)

# Train the model
logistic_model.fit(X_train, y_train)

print("Logistic Regression training completed!")
```

```python
# Predict class labels
y_pred = logistic_model.predict(X_test)

# Predict failure probabilities
y_proba = logistic_model.predict_proba(X_test)[:, 1]

print("First 10 actual labels:")
print(y_test.iloc[:10].tolist())

print("\nFirst 10 predicted labels:")
print(y_pred[:10])

print("\nFirst 10 failure probabilities:")
print(y_proba[:10].round(4))
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

# Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)
pr_auc = average_precision_score(y_test, y_proba)

print("Accuracy:", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall:", round(recall, 4))
print("F1-score:", round(f1, 4))
print("Average Precision (PR-AUC summary):", round(pr_auc, 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    zero_division=0
))
```

```python
import pandas as pd

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

thresholds = [0.50, 0.40, 0.30, 0.20, 0.10]

results = []

for threshold in thresholds:

    # Convert probabilities into class predictions
    y_threshold_pred = (y_proba >= threshold).astype(int)

    # Calculate metrics
    precision = precision_score(
        y_test,
        y_threshold_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_threshold_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        y_threshold_pred,
        zero_division=0
    )

    tn, fp, fn, tp = confusion_matrix(
        y_test,
        y_threshold_pred
    ).ravel()

    results.append({
        "Threshold": threshold,
        "Precision": precision,
        "Recall": recall,
        "F1-score": f1,
        "False Positives": fp,
        "False Negatives": fn,
        "True Positives": tp
    })

threshold_results = pd.DataFrame(results)

display(threshold_results.round(4))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    average_precision_score,
    accuracy_score
)

# Step 1: Define numerical and categorical features
numeric_features = [
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]"
]

categorical_features = ["Type"]

# Step 2: Create preprocessing pipeline
preprocessor_final = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

# Step 3: Create the Random Forest pipeline
rf_model_final = Pipeline(
    steps=[
        ("preprocessor", preprocessor_final),
        (
            "classifier",
            RandomForestClassifier(
                n_estimators=300,
                class_weight="balanced",
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)

# Step 4: Train the model
rf_model_final.fit(X_train_final, y_train_final)

# Step 5: Predict validation probabilities
y_val_proba = rf_model_final.predict_proba(X_val)[:, 1]

# Step 6: Evaluate at the default threshold (0.50)
y_val_pred = (y_val_proba >= 0.50).astype(int)

print("Validation Accuracy:",
      round(accuracy_score(y_val, y_val_pred), 4))

print("\nValidation Average Precision:",
      round(average_precision_score(y_val, y_val_proba), 4))

print("\nClassification Report:")
print(classification_report(y_val, y_val_pred, zero_division=0))

print("Confusion Matrix:")
print(confusion_matrix(y_val, y_val_pred))
```

```python
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
import pandas as pd

thresholds = [0.50, 0.40, 0.30, 0.20, 0.10]

results = []

for threshold in thresholds:

    # Convert probabilities into binary predictions
    y_pred_threshold = (
        y_val_proba >= threshold
    ).astype(int)

    # Calculate confusion matrix
    tn, fp, fn, tp = confusion_matrix(
        y_val,
        y_pred_threshold,
        labels=[0, 1]
    ).ravel()

    # Calculate evaluation metrics
    precision = precision_score(
        y_val,
        y_pred_threshold,
        zero_division=0
    )

    recall = recall_score(
        y_val,
        y_pred_threshold,
        zero_division=0
    )

    f1 = f1_score(
        y_val,
        y_pred_threshold,
        zero_division=0
    )

    results.append({
        "Threshold": threshold,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "True Positives": tp,
        "False Positives": fp,
        "False Negatives": fn
    })

threshold_results = pd.DataFrame(results)

display(
    threshold_results.round(4)
)
```

```python
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from scipy.stats import randint

# Step 1: Define the cross-validation strategy
cv_strategy = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Step 2: Define the hyperparameter search space
param_distributions = {
    "classifier__n_estimators": randint(200, 601),
    "classifier__max_depth": [None, 10, 15, 20, 25, 30],
    "classifier__min_samples_split": randint(2, 21),
    "classifier__min_samples_leaf": randint(1, 11),
    "classifier__max_features": ["sqrt", "log2", 0.8],
    "classifier__class_weight": ["balanced", "balanced_subsample"]
}

# Step 3: Create a fresh pipeline
rf_search_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor_final),
        (
            "classifier",
            RandomForestClassifier(
                random_state=42,
                n_jobs=1
            )
        )
    ]
)

# Step 4: Set up randomized search
rf_search = RandomizedSearchCV(
    estimator=rf_search_pipeline,
    param_distributions=param_distributions,
    n_iter=20,
    scoring="average_precision",
    cv=cv_strategy,
    random_state=42,
    n_jobs=-1,
    verbose=1,
    refit=True
)

# Step 5: Fit using ONLY the training set
rf_search.fit(X_train_final, y_train_final)

# Step 6: Display the best results
print("Best Cross-Validation Average Precision:")
print(round(rf_search.best_score_, 4))

print("\nBest Hyperparameters:")
print(rf_search.best_params_)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

# Step 1: Get the best tuned model
rf_tuned = rf_search.best_estimator_

# Step 2: Predict failure probabilities on validation data
rf_tuned_val_proba = rf_tuned.predict_proba(X_val)[:, 1]

# Step 3: Predict using the default threshold of 0.50
rf_tuned_val_pred = (
    rf_tuned_val_proba >= 0.50
).astype(int)

# Step 4: Evaluate the model
print("Tuned Random Forest — Validation Results\n")

print("Accuracy:",
      round(accuracy_score(y_val, rf_tuned_val_pred), 4))

print("Precision:",
      round(precision_score(
          y_val, rf_tuned_val_pred, zero_division=0
      ), 4))

print("Recall:",
      round(recall_score(
          y_val, rf_tuned_val_pred, zero_division=0
      ), 4))

print("F1-score:",
      round(f1_score(
          y_val, rf_tuned_val_pred, zero_division=0
      ), 4))

print("Average Precision:",
      round(average_precision_score(
          y_val, rf_tuned_val_proba
      ), 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_val, rf_tuned_val_pred))

print("\nClassification Report:")
print(classification_report(
    y_val, rf_tuned_val_pred, zero_division=0
))
```

This project developed a machine learning-based predictive maintenance
system using the UCI AI4I 2020 Predictive Maintenance dataset.

Multiple classification algorithms were evaluated, including Logistic
Regression, Decision Tree, Random Forest, XGBoost, CatBoost, and LightGBM.

- Average Precision: 89.85%

Feature importance and SHAP analysis were used to investigate how
machine attributes contributed to model predictions.

Tool wear, torque, rotational speed, and temperature-related features
were among the important features identified by the model.

1. The AI4I 2020 dataset is synthetic and does not represent all
   real-world industrial operating conditions.
2. The held-out test set contained 51 failure cases, so the estimated
   recall and precision have uncertainty.
3. SHAP explanations describe model behavior and do not establish
   causal relationships.
4. Performance on real industrial data has not been established.
5. External validation is required before practical deployment.
